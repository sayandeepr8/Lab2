Problem 2: The Water Jug Problem (Solved using BFS)
Problem Statement:
You are given two jugs: Jug A with a capacity of 4 litres, and Jug B with a capacity of 3 litres. Neither jug has any markings on it. Using only the following operations — (i) fill a jug completely, (ii) empty a jug completely, (iii) pour water from one jug into the other until either the first jug is empty or the second jug is full — determine the minimum sequence of steps needed to measure out exactly 2 litres of water in any one jug.

```python
from collections import deque

# Capacities of the two jugs
CAP_A = 4
CAP_B = 3


def get_next_states(state):
    a, b = state
    states = []

    # 1. Fill Jug A completely
    states.append(((CAP_A, b), "Fill Jug A"))

    # 2. Fill Jug B completely
    states.append(((a, CAP_B), "Fill Jug B"))

    # 3. Empty Jug A
    states.append(((0, b), "Empty Jug A"))

    # 4. Empty Jug B
    states.append(((a, 0), "Empty Jug B"))

    # 5. Pour Jug A -> Jug B
    amount = min(a, CAP_B - b)
    states.append(((a - amount, b + amount), "Pour A -> B"))

    # 6. Pour Jug B -> Jug A
    amount = min(b, CAP_A - a)
    states.append(((a + amount, b - amount), "Pour B -> A"))

    return states


def bfs():
    start = (0, 0)

    # Queue stores: (current state, path)
    queue = deque()
    queue.append((start, []))

    # Keep track of visited states
    visited = set()
    visited.add(start)

    while queue:
        state, path = queue.popleft()

        a, b = state

        # Goal: exactly 2 litres in either jug
        if a == 2 or b == 2:
            return path + [state]

        # Generate all possible next states
        for next_state, operation in get_next_states(state):

            if next_state not in visited:
                visited.add(next_state)

                new_path = path + [
                    (state, operation)
                ]

                queue.append((next_state, new_path))

    return None


# Solve the problem
solution = bfs()

# Print the solution
print("Minimum sequence of steps:\n")

for i in range(len(solution) - 1):
    state, operation = solution[i]
    next_state = solution[i + 1][0] if isinstance(solution[i + 1], tuple) else solution[i + 1]

    print(f"Step {i + 1}: {operation}")
    print(f"        State: {next_state}")

print("\nGoal reached!")
```
